**CHALLENGE**

```python
while True:
    user_input = input("Enter a number: ")

    if user_input.lower() == "done":
        break

    try:
        num = float(user_input)
        total += num
        count += 1
    except:
        print("Invalid input")
        continue

if count > 0:
    average = total / count
else:
    average = 0

print(total, count, average)
```
