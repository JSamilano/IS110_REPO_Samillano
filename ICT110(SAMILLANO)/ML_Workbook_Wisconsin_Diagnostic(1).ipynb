> **Your answer:**
1.Class Proportions (62.7% Benign vs. 37.3% Malignant): Since there is a mild-to-moderate class imbalance, relying solely on Accuracy can be misleading (a baseline classifier predicting everyone as benign would achieve 62.7% accuracy). In a medical context, missing a malignant case (False Negative) is highly critical. Therefore, we should prioritize metrics like Recall (Sensitivity) for the malignant class (0), F1-Score, and ROC-AUC to measure discrimination.

2.Skewness (Values up to 5.4, e.g., 'area error' and 'concavity error'): Highly skewed features indicate non-normal distributions, with long right tails. Algorithms sensitive to feature distributions (such as Logistic Regression, LDA, or SVMs) benefit greatly from power/logarithmic transforms or standard scaling to normalize distributions and mitigate the leverage of outliers.

**Exercise D.1.** Define the six candidate models: Logistic Regression, LDA, KNN, CART, Gaussian Naive Bayes and SVM.

```python
seed = 7
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)

# TODO: dictionary named models with keys LR, LDA, KNN, CART, NB, SVM
...


def spot_check(models, X_tr, scale):
    """Cross-validated accuracy for every model (optionally scaled inside the pipeline)."""
    results = {}
    for name, model in models.items():
        steps = ([("scale", StandardScaler())] if scale else []) + [("model", model)]
        results[name] = cross_val_score(Pipeline(steps), X_tr, y_train,
                                        cv=kfold, scoring="accuracy")
    return results
```
