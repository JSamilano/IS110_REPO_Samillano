```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
import numpy as np

# ROYGBIV colors with example usage data (you can change these)
colors = ['Red', 'Orange', 'Yellow', 'Green', 'Blue', 'Indigo', 'Violet']
initial_usage = [25, 18, 12, 22, 30, 8, 15]   # Example percentages

hex_colors = ['#FF0000', '#FF7F00', '#FFFF00', '#00FF00', '#0000FF', '#4B0082', '#8F00FF']

df = pd.DataFrame({
    'Color': colors,
    'Usage (%)': initial_usage,
    'Hex': hex_colors
})

# Initialize the Dash app with dark theme
app = dash.Dash(__name__, title="ROYGBIV Color Dashboard")

app.layout = html.Div(style={
    'backgroundColor': '#0a0a0a',
    'color': '#ffffff',
    'fontFamily': 'Arial, sans-serif',
    'padding': '30px',
    'minHeight': '100vh'
}, children=[

    html.H1("🌈 ROYGBIV Color Usage Dashboard",
            style={'textAlign': 'center', 'marginBottom': '10px', 'color': '#ffffff'}),

    html.P("Which color in the rainbow (Red-Orange-Yellow-Green-Blue-Indigo-Violet) is used the most?",
           style={'textAlign': 'center', 'fontSize': '20px', 'marginBottom': '40px', 'color': '#aaaaaa'}),

    # Charts row
    html.Div([
        # Bar Chart
        html.Div([
            dcc.Graph(id='bar-chart')
        ], style={'width': '58%', 'display': 'inline-block', 'paddingRight': '15px'}),

        # Pie Chart
        html.Div([
            dcc.Graph(id='pie-chart')
        ], style={'width': '40%', 'display': 'inline-block', 'verticalAlign': 'top'})
    ], style={'display': 'flex', 'justifyContent': 'center'}),

    # Most Used Color Highlight
    html.Div([
        html.H2("Most Used Color", style={'textAlign': 'center', 'marginTop': '40px'}),
        html.Div(id='most-used-display', style={
            'textAlign': 'center',
            'fontSize': '72px',
            'fontWeight': 'bold',
            'margin': '30px 0',
            'padding': '30px',
            'borderRadius': '15px',
            'backgroundColor': '#1f1f1f',
            'boxShadow': '0 10px 30px rgba(0,0,0,0.6)'
        })
    ]),

    # Interactive Sliders to adjust usage
    html.Div([
        html.H3("Adjust Usage Values (Interactive Simulation)",
                style={'textAlign': 'center', 'marginBottom': '25px'}),

        html.Div([
            html.Label("Red", style={'color': '#FF0000', 'fontWeight': 'bold'}),
            dcc.Slider(id='slider-red', min=0, max=60, step=1, value=25,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),

            html.Label("Orange", style={'color': '#FF7F00', 'fontWeight': 'bold', 'marginTop': '15px'}),
            dcc.Slider(id='slider-orange', min=0, max=60, step=1, value=18,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),

            html.Label("Yellow", style={'color': '#FFFF00', 'fontWeight': 'bold', 'marginTop': '15px'}),
            dcc.Slider(id='slider-yellow', min=0, max=60, step=1, value=12,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),

            html.Label("Green", style={'color': '#00FF00', 'fontWeight': 'bold', 'marginTop': '15px'}),
            dcc.Slider(id='slider-green', min=0, max=60, step=1, value=22,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),

            html.Label("Blue", style={'color': '#0000FF', 'fontWeight': 'bold', 'marginTop': '15px'}),
            dcc.Slider(id='slider-blue', min=0, max=60, step=1, value=30,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),

            html.Label("Indigo", style={'color': '#4B0082', 'fontWeight': 'bold', 'marginTop': '15px'}),
            dcc.Slider(id='slider-indigo', min=0, max=60, step=1, value=8,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),

            html.Label("Violet", style={'color': '#8F00FF', 'fontWeight': 'bold', 'marginTop': '15px'}),
            dcc.Slider(id='slider-violet', min=0, max=60, step=1, value=15,
                       marks={0: '0%', 30: '30%', 60: '60%'}, tooltip={"placement": "bottom"}),
        ], style={'maxWidth': '700px', 'margin': '0 auto'})
    ], style={'marginTop': '50px', 'padding': '30px', 'backgroundColor': '#1a1a1a', 'borderRadius': '15px'})
])

# Callback to update everything live
@app.callback(
    [Output('bar-chart', 'figure'),
     Output('pie-chart', 'figure'),
     Output('most-used-display', 'children')],
    [Input('slider-red', 'value'),
     Input('slider-orange', 'value'),
     Input('slider-yellow', 'value'),
     Input('slider-green', 'value'),
     Input('slider-blue', 'value'),
     Input('slider-indigo', 'value'),
     Input('slider-violet', 'value')]
)
def update_dashboard(r, o, y, g, b, i, v):
    usage = [r, o, y, g, b, i, v]
    df_updated = pd.DataFrame({
        'Color': colors,
        'Usage (%)': usage,
        'Hex': hex_colors
    })

    # Bar Chart
    bar_fig = px.bar(df_updated, x='Color', y='Usage (%)',
                     color='Color',
                     color_discrete_sequence=hex_colors,
                     text='Usage (%)')
    bar_fig.update_layout(
        title="Usage Frequency by Color",
        plot_bgcolor='#1a1a1a',
        paper_bgcolor='#0a0a0a',
        font_color='#ffffff',
        title_font_size=20,
        xaxis_title="Color",
        yaxis_title="Usage (%)",
        bargap=0.2
    )

    # Pie Chart
    pie_fig = px.pie(df_updated, names='Color', values='Usage (%)',
                     color='Color', color_discrete_sequence=hex_colors,
                     hole=0.45)
    pie_fig.update_layout(
        title="Color Distribution",
        plot_bgcolor='#1a1a1a',
        paper_bgcolor='#0a0a0a',
        font_color='#ffffff',
        title_font_size=20
    )

    # Find the most used color
    max_idx = np.argmax(usage)
    most_color = colors[max_idx]
    most_hex = hex_colors[max_idx]
    most_value = usage[max_idx]

    most_used = html.Div([
        html.Span(most_color, style={
            'color': most_hex,
            'fontSize': '90px',
            'textShadow': f'0 0 20px {most_hex}'
        }),
        html.Br(),
        html.Span(f"{most_value}%", style={'fontSize': '36px', 'color': '#aaaaaa'})
    ])

    return bar_fig, pie_fig, most_used

# Run the app
if __name__ == '__main__':
    print("🚀 Starting ROYGBIV Color Dashboard...")
    print("Open your browser at: http://127.0.0.1:8050/")
    app.run(debug=True)
```
